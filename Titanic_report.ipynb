### This report explains how the chosen visualisations inform feature selection

## Approach
I reviewed the existing exploratory plots and kept those that most clearly separate survivors from non-survivors. The aim is to decide which features to carry into modelling. All figures are saved in `figures/`.


## 1. Sex and passenger class: strongest candidates

![Passengers by sex](figures/number_of_passengers_sex.png)
![Passengers by class](figures/number_of_passengers_pclass.png)

Female passengers survived at a much higher rate than male passengers, and first-class passengers fared better than those in
lower classes, with the advantage largest for first-class women.
Both features show a large, clear gap between survivors and non-survivors,
so **Sex** and **Pclass** should be included. The visible interaction
between them suggests that a model able to capture interactions
(or an explicit Sex × Pclass feature) may help.

## 2. Age and fare: informative continuous features

![Distribution of age](figures/distribution_of_age.png)
![Distribution of fare](figures/distribution_of_fare.png)

The distributions differ by outcome. Younger passengers were more likely to
survive, and passengers who paid higher fares were also more likely to
survive. Both support including **Age** and **Fare**. Fare is likely
partly a proxy for class, so the two are correlated; this should be
checked (e.g. with a correlation matrix) before modelling. Fare is also
heavily left-skewed, so a log transform or binning may be useful.
Age needs imputation for missing values.

## 3. Family size: useful but ambiguous

![Passengers by family size](figures/number_of_passengers_family_size.png)
![Passengers by SibSp](figures/number_of_passengers_sibsp.png)
![Passengers by Parch](figures/number_of_passengers_parch.png)

Passengers travelling in small to medium families (2-3 members) were more
likely to survive than not, while those travelling alone or in large
families were less likely to. One possible explanation is that many of
these passengers were children travelling with parents, who were prioritised
for lifeboats. This is only a hypothesis: the chart does not control for
age, sex or class, so the effect may reflect those variables instead.
Because the combined **family size** feature summarises SibSp and Parch
in one variable and shows a clearer pattern, I would keep it and treat
SibSp and Parch individually as lower priority to avoid redundancy.

## 4. Embarked: weaker evidence

![Passengers by port of embarkation](figures/number_of_passengers_embarked.png)

Any difference by port may reflect class mix at each port rather than a direct effect, so Embarked
is a lower-priority feature that I would test in the model rather than assume.

## Summary

| Feature | Evidence | Decision |
|---|---|---|
| Sex | Strong | Include |
| Pclass | Strong | Include |
| Age | Clear | Include (impute missing) |
| Fare | Clear, correlated with class | Include (transform) |
| Family size | Moderate, confounded | Include, test |
| SibSp / Parch | Redundant with family size | Lower priority |
| Embarked | [Weak / to check] | Test in model |

**Limitation:** these charts show marginal relationships only. Confounding
between variables (class, sex, age, fare) means modelling is needed to
separate their effects.
